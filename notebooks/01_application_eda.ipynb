# Notebook 1 - Application Train EDA

This notebook explores the main application file (application_train.csv).
The goal is to understand the data structure, identify missing values, 
spot anomalies, and find which features are most related to default risk 
before any modelling begins.

## Import libraries


```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
pd.set_option('display.max_rows', None)

print("Libraries loaded.")
```

## Load the data


```python
application = pd.read_csv("data/raw/application_train.csv")
application.shape
```

```python
application.head()
```

## Understand the dataset


```python
numeric_columns = application.select_dtypes(include="number").columns.tolist()
categorical_columns = application.select_dtypes(exclude="number").columns.tolist()

print("Numeric columns:", len(numeric_columns))
print("Categorical columns:", len(categorical_columns))
print("Exact duplicate rows:", application.duplicated().sum())
print("Duplicate applicant IDs:", application["SK_ID_CURR"].duplicated().sum())
print("Missing applicant IDs:", application["SK_ID_CURR"].isna().sum())
print("Target values:", sorted(application["TARGET"].dropna().unique()))
```

```python
data_type_summary = application.dtypes.astype(str).value_counts().rename_axis("data_type").reset_index(name="column_count")
data_type_summary
```

## Target distribution


```python
target_count = application["TARGET"].value_counts().sort_index()
target_percent = application["TARGET"].value_counts(normalize=True).sort_index() * 100

target_summary = pd.DataFrame({
    "count": target_count,
    "percent": target_percent.round(2)
})
target_summary.index = ["No default (0)", "Default (1)"]
display(target_summary)



plt.figure(figsize=(5, 4))
sns.countplot(data=application, x="TARGET")
plt.title("Default Distribution")
plt.xlabel("Default (0 = No, 1 = Yes)")
plt.ylabel("Number of applicants")
plt.tight_layout()
plt.show()
```

The target is imbalanced. Only 8.07% of applicants defaulted.


## Missing values


```python
missing_percent = (application.isnull().mean() * 100).round(2)
missing_percent = missing_percent[missing_percent > 0].sort_values(ascending=False)

missing_summary = missing_percent.to_frame(name="missing_percent")

print("Columns with missing values:", len(missing_summary))
display(missing_summary)
```

The amount of missing data is different for each column, so one rule will not work for all of them. Each column needs to be looked at before deciding what to keep or drop.


## Missing-value groups


```python
high_missing = missing_percent[missing_percent > 50]
medium_missing = missing_percent[(missing_percent > 5) & (missing_percent <= 50)]
low_missing = missing_percent[missing_percent <= 5]

print("More than 50% missing:", len(high_missing))
print("Between 5% and 50%:", len(medium_missing))
print("Up to 5% missing:", len(low_missing))
```

Most of the columns with a lot of missing values are about buildings or housing. What to drop will be decided after checking which columns actually matter for default.


### Top 30 columns with missing values


```python
top_missing = missing_percent.head(30).sort_values()

plt.figure(figsize=(10, 8))
plt.barh(top_missing.index, top_missing.values)
plt.axvline(50, color="red", linestyle="--", label="50% missing")
plt.title("Top 30 Columns with Missing Values")
plt.xlabel("Missing values (%)")
plt.legend()
plt.tight_layout()
plt.show()
```

Most of these columns are about buildings or housing. EXT_SOURCE_1 also has a lot of missing values, but it looks useful, so it will be kept.


## Special and unusual values


```python
days_employed_sentinel = (application["DAYS_EMPLOYED"] == 365243).sum()
days_employed_sentinel_percent = days_employed_sentinel / len(application) * 100

print("DAYS_EMPLOYED equal to 365243:", days_employed_sentinel)
print("Percentage:", round(days_employed_sentinel_percent, 2), "%")
print()

print(application["CODE_GENDER"].value_counts(dropna=False))
print()
print("ORGANIZATION_TYPE values containing XNA:")
print((application["ORGANIZATION_TYPE"] == "XNA").sum())
```

DAYS_EMPLOYED has the value 365243 for 18% of applicants. This is not a real number of days worked, it is a placeholder for "not employed." CODE_GENDER has 4 XNA values, and ORGANIZATION_TYPE also has an XNA category. These will be treated as unknown during cleaning.


### Basic data checks

Checking a few simple rules. Amounts and counts should not be negative, age is stored as negative days, and the target should always be present.


```python
domain_checks = pd.DataFrame({
    "check": [
        "DAYS_BIRTH is zero or positive",
        "AMT_INCOME_TOTAL is zero or negative",
        "AMT_CREDIT is zero or negative",
        "AMT_ANNUITY is zero or negative",
        "CNT_CHILDREN is negative",
        "CNT_FAM_MEMBERS is zero or negative",
        "TARGET is missing"
    ],
    "rows_affected": [
        (application["DAYS_BIRTH"] >= 0).sum(),
        (application["AMT_INCOME_TOTAL"] <= 0).sum(),
        (application["AMT_CREDIT"] <= 0).sum(),
        (application["AMT_ANNUITY"] <= 0).sum(),
        (application["CNT_CHILDREN"] < 0).sum(),
        (application["CNT_FAM_MEMBERS"] <= 0).sum(),
        application["TARGET"].isna().sum()
    ]
})
domain_checks
```

## Numerical summaries and outliers


```python
analysis_numeric_columns = [column for column in numeric_columns if column not in ["SK_ID_CURR", "TARGET"]]

numeric_summary = application[analysis_numeric_columns].describe(
    percentiles=[0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]
).T

numeric_summary[["count", "mean", "std", "min", "1%", "50%", "99%", "max"]].head(30)
```

```python
key_numeric_features = [
    "AMT_INCOME_TOTAL", "AMT_CREDIT", "AMT_ANNUITY",
    "CNT_CHILDREN", "CNT_FAM_MEMBERS",
    "OBS_30_CNT_SOCIAL_CIRCLE", "DEF_30_CNT_SOCIAL_CIRCLE"
]

extreme_value_table = numeric_summary.loc[key_numeric_features, ["min", "1%", "50%", "99%", "max"]]
extreme_value_table
```

Some variables have maximum values that are much higher than their 99th percentile. These are not necessarily errors. They will be looked at again during cleaning.


### Important numerical distributions

Age and employment are converted from days to years, only for this chart. Income and credit are also capped at the 99th percentile just for the chart. The actual data is not changed.


```python
plot_data = application.copy()
plot_data["AGE_YEARS"] = -plot_data["DAYS_BIRTH"] / 365.25
plot_data["EMPLOYMENT_YEARS"] = np.where(
    plot_data["DAYS_EMPLOYED"] == 365243,
    np.nan,
    -plot_data["DAYS_EMPLOYED"] / 365.25
)

income_limit = plot_data["AMT_INCOME_TOTAL"].quantile(0.99)
credit_limit = plot_data["AMT_CREDIT"].quantile(0.99)

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
sns.histplot(data=plot_data, x="AGE_YEARS", hue="TARGET", palette={0: "#4C78A8", 1: "#E45756"}, bins=40, stat="density", common_norm=False, ax=axes[0, 0])
sns.histplot(data=plot_data, x="EMPLOYMENT_YEARS", hue="TARGET", palette={0: "#4C78A8", 1: "#E45756"}, bins=40, stat="density", common_norm=False, ax=axes[0, 1])
sns.histplot(data=plot_data[plot_data["AMT_INCOME_TOTAL"] <= income_limit], x="AMT_INCOME_TOTAL", hue="TARGET", palette={0: "#4C78A8", 1: "#E45756"}, bins=40, stat="density", common_norm=False, ax=axes[1, 0])
sns.histplot(data=plot_data[plot_data["AMT_CREDIT"] <= credit_limit], x="AMT_CREDIT", hue="TARGET", palette={0: "#4C78A8", 1: "#E45756"}, bins=40, stat="density", common_norm=False, ax=axes[1, 1])

axes[0, 0].set_title("Age Distribution by Target")
axes[0, 1].set_title("Employment Years by Target")
axes[1, 0].set_title("Income Distribution by Target (up to 99th percentile)")
axes[1, 1].set_title("Credit Distribution by Target (up to 99th percentile)")
plt.tight_layout()
plt.show()

del plot_data
```

Applicants who defaulted tend to be a little younger and have less employment history, but the two groups overlap a lot. Income and credit are right-skewed and also overlap a lot between the two groups. These features may only be useful when combined with other features, not on their own.


## Numerical relationships with default


```python
correlation_data = application[numeric_columns].copy()
correlation_data["DAYS_EMPLOYED"] = correlation_data["DAYS_EMPLOYED"].replace(365243, np.nan)

target_correlations = correlation_data.corr(numeric_only=True)["TARGET"].drop(["TARGET", "SK_ID_CURR"])
target_correlations = target_correlations.sort_values()

top_correlations = pd.concat([target_correlations.head(12), target_correlations.tail(12)]).drop_duplicates()
top_correlations.to_frame("correlation_with_target")
```

```python
plt.figure(figsize=(9, 8))
top_correlations.sort_values().plot(kind="barh")
plt.axvline(0, color="black", linewidth=1)
plt.title("Strongest Numerical Correlations with Target")
plt.xlabel("Correlation")
plt.tight_layout()
plt.show()
```

EXT_SOURCE_1, EXT_SOURCE_2 and EXT_SOURCE_3 have the strongest correlation with TARGET. The correlation is negative, so a higher score means a lower default risk. Every other variable has a weak correlation on its own.


### Correlation heatmap with target

Only the top 10 correlated variables are shown. Showing all of them together would be too crowded to read.


```python
top_target_features = (
    target_correlations.abs()
    .sort_values(ascending=False)
    .head(10)
    .index
    .tolist()
)

correlation_matrix = correlation_data[top_target_features + ["TARGET"]].corr()

plt.figure(figsize=(11, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)
plt.title("Correlation Matrix of Key Numerical Features and Target")
plt.tight_layout()
plt.show()
```

This confirms that the EXT_SOURCE features matter the most. It also shows that some of the top features are correlated with each other.


### External-source features

Looking at these three separately, since they are the strongest predictors.


```python
external_columns = [
    "EXT_SOURCE_1",
    "EXT_SOURCE_2",
    "EXT_SOURCE_3"
]

external_summary = pd.DataFrame({
    "missing_percent": application[external_columns].isna().mean() * 100,
    "correlation_with_target": application[
        external_columns + ["TARGET"]
    ].corr()["TARGET"].drop("TARGET")
}).round(3)

external_summary
```

```python
application[external_columns].corr().round(3)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for index, column in enumerate(external_columns):
    sns.histplot(
        data=application,
        x=column,
        hue="TARGET",
        palette={0: "#4C78A8", 1: "#E45756"},
        bins=30,
        stat="density",
        common_norm=False,
        ax=axes[index]
    )
    axes[index].set_title(f"{column} by Default Status")

plt.tight_layout()
plt.show()
```

All three EXT_SOURCE variables are negatively correlated with default. They are related to each other but not identical, so it is worth keeping all three, or a combined version of them.


## Categorical relationships with default


```python
cat_features = [
    "CODE_GENDER",
    "NAME_EDUCATION_TYPE",
    "NAME_INCOME_TYPE",
    "NAME_FAMILY_STATUS",
    "NAME_HOUSING_TYPE",
    "NAME_CONTRACT_TYPE",
    "OCCUPATION_TYPE",
    "ORGANIZATION_TYPE"
]

for col in cat_features:
    result = (
        application.groupby(col)["TARGET"]
        .mean() * 100
    ).round(2).sort_values(ascending=False)

    plt.figure(figsize=(9, 4))
    result.plot(kind="bar")
    plt.title(f"Default Rate by {col}")
    plt.ylabel("Default Rate (%)")
    plt.xticks(rotation=25, ha="right")
    plt.tight_layout()
    plt.show()

    print(result.to_string())
    print()
```

Default rates are different across education, income type, housing, occupation, and other groups. Some of these categories have very few applicants, so an unusually high or low rate in a small group should not be trusted too much.


## Main EDA findings

- 307,511 applicants, 122 columns, no duplicates or missing IDs.
- The target is imbalanced: 8.07% defaulted.
- 64 columns have missing values, and 41 of them are missing more than half their data, mostly building and housing information.
- DAYS_EMPLOYED has a placeholder value (365243) that needs to be fixed.
- A few unknown or XNA categories need cleaning up.
- EXT_SOURCE_1, EXT_SOURCE_2 and EXT_SOURCE_3 are the strongest predictors of default.
- Age, employment, income, credit, and several categorical variables show some difference by default status, though with a lot of overlap.

Next step: handle the missing values, fix the placeholder value, and clean up the rare and unknown categories.
