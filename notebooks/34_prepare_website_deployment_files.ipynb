# Prepare Website Deployment Files

This notebook is not part of the modelling pipeline or the academic evaluation, it is a separate, optional step that packages the already-frozen final models into a single zip file for reuse outside this project (for example, a demo application). It only copies files that were already saved by notebook 30, nothing is retrained or recalculated here.


## Set project paths


```python
from pathlib import Path
import shutil

current_folder = Path.cwd().resolve()
project_root = current_folder.parent if current_folder.name == "notebooks" else current_folder

result_folder = project_root / "data" / "modeling" / "final_test_results"
model_folder = result_folder / "trained_models"
```

Same paths notebook 30 used to save its results, this notebook just needs to know where to find the files that are already there.


## Collect and zip the frozen model files


```python
website_files = project_root / "website_model_files"
website_files.mkdir(exist_ok=True)

files_to_copy = [
    model_folder / "final_ordinal_preprocessor.joblib",
    model_folder / "final_xgboost.json",
    model_folder / "final_lightgbm.txt",
    model_folder / "final_catboost.cbm",
    result_folder / "final_predictor_list.csv",
]

for file in files_to_copy:
    assert file.exists(), f"File not found: {file}"
    shutil.copy2(file, website_files / file.name)

zip_path = shutil.make_archive(
    str(project_root / "website_model_files"),
    "zip",
    website_files
)

print("ZIP file created:")
print(zip_path)
```

This collects the frozen preprocessor, the three trained models, and the final predictor list into one folder and zips it up. If any of these files are missing, the assert above will fail with a clear message telling you which one, most likely meaning notebook 30 needs to be run first.
