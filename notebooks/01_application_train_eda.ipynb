# EDA — application_train.csv

## Step 1 — Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
print("Libraries loaded")
```

## Step 2 — Load Data

```python
df = pd.read_csv("data/application_train.csv")

df.columns = df.columns.str.lower().str.strip()

if "target" in df.columns:
    df = df.rename(columns={"target": "default"})

print("Shape:", df.shape)
```

```python
print("DATA TYPES")
print(df.dtypes.value_counts())
print()

numeric_cols = df.select_dtypes(include=np.number).columns.tolist()
numeric_cols = [c for c in numeric_cols if c not in ["sk_id_curr", "default"]]

categorical_cols = df.select_dtypes(exclude=np.number).columns.tolist()

print("Numeric columns    :", len(numeric_cols))
print("Categorical columns:", len(categorical_cols))
print()
print("Categorical columns:")
categorical_cols
```

```python
print("Duplicate rows    :", df.duplicated().sum())
print("Unique applicants :", df["sk_id_curr"].nunique())
```

## Step 4 — Basic Statistics

```python
df[numeric_cols].describe().T.round(2)
```

## Step 5 — Target Variable

```python
print("=== TARGET DISTRIBUTION ===")
print(df["default"].value_counts())
print()
print((df["default"].value_counts(normalize=True) * 100).round(2))
```

```python
plt.figure(figsize=(5, 4))
sns.countplot(x=df["default"])
plt.title("Default Distribution")
plt.xlabel("Default (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Step 6 — Missing Values

```python
missing = (df.isnull().sum() / len(df) * 100).round(2)
missing = missing[missing > 0].sort_values(ascending=False)

print(f"Columns with missing values: {len(missing)} out of {df.shape[1]}")
print()
print(missing)

plt.figure(figsize=(10, 8))
sns.barplot(x=missing.values, y=missing.index)
plt.title("application_train — Missing Value %")
plt.xlabel("Missing %")
plt.tight_layout()
plt.show()
```

```python
high   = missing[missing > 40]
medium = missing[(missing > 5) & (missing <= 40)]
low    = missing[missing <= 5]

print("High   (>40%) :", len(high),   "columns")
print("Medium (5-40%):", len(medium), "columns")
print("Low    (<5%)  :", len(low),    "columns")
print()
print("MEDIUM MISSING (5-40%)")
print(medium)
print()
print("LOW MISSING (<5%)")
print(low)
```

```python
plt.figure(figsize=(10, 8))
sns.barplot(x=missing.head(30).values, y=missing.head(30).index)
plt.axvline(40, color="red",    linestyle="--", label="40% threshold")
plt.axvline(5,  color="orange", linestyle="--", label="5% threshold")
plt.title("Top 30 Columns with Missing Values")
plt.xlabel("Missing %")
plt.legend()
plt.tight_layout()
plt.show()
```

## Step 7 — Outlier Check

```python
flag_cols  = [c for c in numeric_cols if c.startswith("flag_")]
check_cols = [c for c in numeric_cols if c not in flag_cols]

print(f"Checking {len(check_cols)} columns\n")
print(f"{'Column':<40} {'1%':>12} {'99%':>12} {'Max':>15} {'Outlier?'}")
print("-" * 90)

outlier_cols = []
for col in check_cols:
    q01     = df[col].quantile(0.01)
    q99     = df[col].quantile(0.99)
    max_val = df[col].max()
    is_out  = (max_val > q99 * 3) if q99 > 0 else (
               max_val > 100 if q99 == 0 else False)
    flag = "YES" if is_out else "ok"
    if is_out:
        outlier_cols.append(col)
    print(f"{col:<40} {q01:>12.2f} {q99:>12.2f} {max_val:>15.2f}  {flag}")

print()
print("=== COLUMNS WITH EXTREME OUTLIERS ===")
for c in outlier_cols:
    print(" -", c)
```

## Step 9 — Distribution of Key Features

```python
# days_birth is stored as negative days
# We flip the sign and divide by 365 to get age in years
df["age_years"] = (-df["days_birth"]) / 365

# days_employed contains 365243 as a placeholder for unemployed
# Replace with NaN first
df["days_employed"] = df["days_employed"].replace(365243, np.nan)
df["employment_years"] = (-df["days_employed"]) / 365

print("=== AGE ===")
print(df["age_years"].describe().round(2))
print()
print("=== EMPLOYMENT YEARS (after fixing 365243 anomaly) ===")
print(df["employment_years"].describe().round(2))
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.kdeplot(data=df, x="age_years", hue="default", fill=True, ax=axes[0][0])
axes[0][0].set_title("Age Distribution by Default Status")
axes[0][0].set_xlabel("Age (years)")

sns.kdeplot(data=df, x="employment_years", hue="default", fill=True, ax=axes[0][1])
axes[0][1].set_xlim(0, 40)
axes[0][1].set_title("Employment Years by Default Status")
axes[0][1].set_xlabel("Employment Years")

q99_inc = df["amt_income_total"].quantile(0.99)
sns.kdeplot(
    data=df[df["amt_income_total"] <= q99_inc],
    x="amt_income_total", hue="default", fill=True, ax=axes[1][0]
)
axes[1][0].set_title("Income Distribution by Default Status")
axes[1][0].set_xlabel("Total Income")

q99_cred = df["amt_credit"].quantile(0.99)
sns.kdeplot(
    data=df[df["amt_credit"] <= q99_cred],
    x="amt_credit", hue="default", fill=True, ax=axes[1][1]
)
axes[1][1].set_title("Credit Amount by Default Status")
axes[1][1].set_xlabel("Credit Amount")

plt.tight_layout()
plt.show()
```

## Step 10 — Correlation with Target

```python
numeric_df = df.select_dtypes(include=np.number)
corr = numeric_df.corr()["default"].drop("default")

print("=== TOP 15 POSITIVE CORRELATIONS ===")
print(corr.sort_values(ascending=False).head(15).round(4))
print()
print("=== TOP 15 NEGATIVE CORRELATIONS ===")
print(corr.sort_values().head(15).round(4))
```

## Step 11 — EXT_SOURCE Features

```python
ext_cols = ["ext_source_1", "ext_source_2", "ext_source_3"]

print("=== MISSING VALUES ===")
for col in ext_cols:
    pct = df[col].isnull().mean() * 100
    print(f"  {col}: {pct:.2f}% missing")

print()
print("=== INTER-CORRELATION ===")
print(df[ext_cols].corr().round(3))
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))
for i, col in enumerate(ext_cols):
    sns.kdeplot(data=df, x=col, hue="default", fill=True, ax=axes[i])
    axes[i].set_title(f"{col.upper()} by Default Status")
plt.tight_layout()
plt.show()
```

## Step 12 — Relationship with Target (Categorical Features)

```python
cat_features = [
    "code_gender",
    "name_education_type",
    "name_income_type",
    "name_family_status",
    "name_housing_type",
    "name_contract_type",
    "occupation_type",
    "organization_type"
]

for col in cat_features:
    result = (
        df.groupby(col)["default"]
        .mean() * 100
    ).round(2).sort_values(ascending=False)

    plt.figure(figsize=(9, 4))
    result.plot(kind="bar")
    plt.title(f"Default Rate by {col}")
    plt.ylabel("Default Rate (%)")
    plt.xticks(rotation=25, ha="right")
    plt.tight_layout()
    plt.show()

    print(result.to_string())
    print()
```

```python
top_corr = pd.concat([
    corr.sort_values(ascending=False).head(10),
    corr.sort_values().head(10)
]).drop_duplicates()

plt.figure(figsize=(10, 6))
top_corr.sort_values().plot(kind="barh")
plt.title("Top Feature Correlations with Default")
plt.xlabel("Correlation")
plt.axvline(0, color="black", linewidth=0.8)
plt.tight_layout()
plt.show()
```

## Step 13 — Correlation Matrix

```python
# Selecting the most relevant columns for the correlation matrix
# Using top correlated features — full matrix would be unreadable

corr_cols = [
    "ext_source_1", "ext_source_2", "ext_source_3",
    "age_years", "employment_years",
    "amt_income_total", "amt_credit", "amt_annuity",
    "region_rating_client_w_city", "days_last_phone_change",
    "def_30_cnt_social_circle", "def_60_cnt_social_circle",
    "default"
]

corr_matrix = df[corr_cols].corr().round(2)

plt.figure(figsize=(12, 10))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    vmin=-1,
    vmax=1,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8}
)

plt.title("Correlation Matrix — Key Features", fontsize=14, pad=15)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

## Observations

- 307,511 applicants, 122 columns, no duplicate rows found.
- Target variable is heavily imbalanced: 91.93% non-default vs 8.07% default.
- 64 out of 122 columns have missing values.
  - 49 columns have more than 40% missing — these will be dropped.
  - 8 columns have 5-40% missing — these will be imputed carefully.
  - 7 columns have less than 5% missing — simple imputation.
- days_employed contains 365243 as a placeholder for unemployed applicants — replaced with NaN.
- flag_mobil has zero variance — every applicant has value 1 — useless column, will be dropped.
- amt_income_total max is 117,000,000 vs 99th percentile 472,500 — extreme outlier, cap required.
- cnt_children max is 19 — data entry error, cap at 5.
- obs_30_cnt_social_circle max is 348 — impossible value, cap required.
- EXT_SOURCE features are the strongest predictors of default (correlations -0.155 to -0.179).
- EXT_SOURCE inter-correlations are low (0.109-0.214) confirming they are independent sources.
- Lower education level is strongly associated with higher default rate (10.93% vs 1.83%).
- Rented accommodation applicants default more than property owners (12.31% vs 7.80%).
- Males default more than females (10.14% vs 7.00%).
- Low-skill laborers have the highest default rate by occupation (17.15%).
